```python
# Target distribution 

sns.histplot(df['productivity_score'], bins=30, kde=True)
plt.title("Productivity Score Distribution")
plt.tight_layout()
plt.show()
```

```python
# boxplot

sns.boxplot(df['productivity_score'])
plt.title("Boxplot of target")
plt.tight_layout()
plt.show()
```

```python
# features distribution

numeric_features=[ 'age', 'experience_years','technical_skill_score',
                         'communication_score','avg_daily_focus_hours',
                         'remote_work_ratio','sleep_hours'

]
```

```python
categorical_features = [
     "education_level",
    "stress_level",
    "project_complexity",
    'team_size',
    "manager_rating"

]
```

```python
# univarient analysis

df[numeric_features].hist(
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
# categorical

for col in categorical_features:
    print(f"Value counts of {col} :\n")
    print(df[col].value_counts())
```

```python
for col in categorical_features:
    sns.countplot(data=df, x=col)
    plt.title(f"Distribution of {col}")
    plt.tight_layout()
    plt.show()

```

```python
# bivarient 

# target Vs numeric


for col in numeric_features:
    sns.scatterplot(data=df, x=col, y="productivity_score")
    plt.title(f"{col} Vs Productivity Score")
    plt.tight_layout()
    plt.show()
```

```python
# target vs categorical

for col in categorical_features:
    sns.boxplot(data=df, x=col, y="productivity_score")
    plt.title(f"{col} Vs Productivity Score")
    plt.tight_layout()
    plt.show()
```

```python
correlation_matrix=df.corr(numeric_only=True)

plt.figure(figsize=(16,14))

sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm')

plt.title("Correlation matrix")
plt.tight_layout()
plt.show()
```

```python
# correlation with target

target_correlation = (
    df.corr(numeric_only=True)['productivity_score'].sort_values(ascending=False)

)

target_correlation
```
